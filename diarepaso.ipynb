# Challenge 🤺 

Define una variable que contenga un numero secreto. Crea un programa que teniendo una cantidad limitada de intentos, preguntar al usuario por el numero hasta que acierte. Por cada intento deberas:
- Si falla imprimir si el numero a adivinar es mayor o menor que el numero ingresado 
- indicar siempre la cantidad de intentos que le quedan.
- Si acierta imprimir un mensaje.

```python
# definir una variable con un numero secreto
numero_secreto = 7

# crear un programa que: 
# con numeros limitados de intentos
intento = 5
while intento > 0:
    intento = intento - 1

    # consulte al usuario un numero. hasta que acierte.
    numero_ingresado = int(input("Ingrese un numero"))

    # si falla, mostrar si el numero es mayor o menor
    # siempre indicar la cantidad de intentos
    if numero_ingresado < numero_secreto:
        print("Numero ingresado es menor al numero secreto")
        print(f"Te quedan {intento} intentos")
    elif numero_ingresado > numero_secreto:
        print("Numero ingresado es mayor al numero secreto")
        print(f"Te quedan {intento} intentos")
    elif intento == 0:
        print("Te quedaste sin intentos")     
    # si acierta, imprimir un mensaje
    else:
        print("Felicidades!! adivinaste")
        break
```

- mensaje de felicitaciones si acerto

# CUIDADO CON LOS BUCLES INFINITOS!!

```python
# Ivan

palabra_secreta = "python"

max_intentos =  3

intentos = 0



while intentos < max_intentos:

    intento = input("Adivina la palabra secreta: ")
    intentos += 1

    if intento.lower() == palabra_secreta:
        print("Felicidades aivinaste la palabra secreta!!!")
        break
    else: 
        print("Incorrecto. intenta de nuevo.")

if intento.lower() != palabra_secreta:
    print("se acabaron los intentos. La palabra secreta era: ", palabra_secreta)
```

# Return

```python
def sumar_numeros(num1, num2):
    sumar = num1 + num2
    multi = num1 * num2
    return sumar, multi

numero_1 = int(input("ingrese un numero"))
numero_2 = int(input("ingrese otro numero"))


resultado_suma, resultado_multi = sumar_numeros(numero_1, numero_2)
print(resultado_suma)
print(resultado_multi)

```
