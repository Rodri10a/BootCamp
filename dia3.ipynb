## CHALLENGE


Crear función que revise si una palabra contiene una letra. En caso de que la contenga, retornar la letra entre corchetes []. De lo contrario, retornar solo la letra

```python
def retorno (word,letra):
    for i in word:
        if letra in i: 
            return (f"[{letra}]")  

        else:   
            return letra
        
print()
```
