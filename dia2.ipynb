## CHALLENGE 

Escribir un programa en el que se pregunte al usuario por una temperatura (en grados Celsius), imprima la temperatura escrita, y luego clasificar la temperatura ingresada en los siguientes niveles:

Entre -10 y 18 grados, que imprima "Fresquete".
Entre 19 y 30 grados, que imprima "Calor'i".
Entre 31 y 45 grados, que imprima "Hakuuuuuu".
Para todo lo demás, imprimir "Ñamano mba'e!".


```python
def clasificador (tem) : 
    if -10 <= tem <= 18 :    #crear una condicional if para saber si esta entre -10 a 18 grados 
        print ("Fresquetee")  #imprimir este resultado si se cumple la condicinal if 
    elif 19 <= tem <=  30:  # crear una condicional elif para saber si esta entre 19 a 30 grados 
        print ("Calor'i ")  # imprimir este mensaje si coincide con este primer elif
    elif 31 <= tem <= 45:  # crear otra  condicional elif para saber si esta entre 31 a 45 grados
        print ("Hakuuu")  # imprimir este mensaje si coincide con este segundo elif 
    else:   # crear una condicional else para que se ejecute si ninguna de tus condicionales anteriores se cumplio 
        print("Ñamano mba'e")   #imprimir este codigo si ninguna de las anteriores opciones coinciden        
tem = int(input("Ingrese la temperatura actual en grados Celsius")) #pedirle al usuario la temperatura en grados Celsius 
print (f"La temperatura ahora es de {tem}") #imprimir el resultado que te dio el usuario 
clasificador(tem) 
```

## BUCLE CONDICIONAL (WHILE)

```python
num_adivinar = 20 
numero = 0 

while numero != num_adivinar: # el bucle solo se rompe si las variables son iguales porque usaste (!=)
    numero = int(input("Ingrese un numero"))
    print (numero) 
    if numero != num_adivinar:
        print ("Sorry, try again")
print ('Adivinaste!!')
```

## LISTAS 

```python
super = ['manzana', 'naranja', 'pera','banana']
dif = [True, 2 , 'hola', 2.1] #puede ser compuesta igualmente
num = [1,2,3,4,5,6,7,8,9]  #pueden haber lista flotantes, booleanas,string

# Acceder a un elemento de la lista
print (super[3])
print (dif[0])
print (num[5])

# Agregar un elemento a la lista

super.append ('uva')
print(super)

#Modificar elementos
super[1] = 'pinha'
print(super)

#eliminar un elemento especifico
super.remove ('manzana')
print (super)

#modificar el ultimo elemento 
super.pop (2)

#contar los elementos de tu lista 
print(len(super)) 
```
